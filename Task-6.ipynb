```python
import matplotlib.pyplot as plt

# Data
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

# Create line chart
plt.plot(x, y, marker='o')

# Labels and title
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Line Chart")

# Grid
plt.grid(True)

# Show graph
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
students = ['A', 'B', 'C', 'D']
marks = [85, 90, 78, 88]

# Create bar chart
plt.bar(students, marks)

# Labels and title
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Bar Chart")

# Show graph
plt.show()
```

```python
import matplotlib.pyplot as plt

# Data
hours = [1, 2, 3, 4, 5, 6]
scores = [35, 40, 50, 65, 70, 90]

# Create scatter plot
plt.scatter(hours, scores)

# Labels and title
plt.xlabel("Study Hours")
plt.ylabel("Scores")
plt.title("Scatter Plot")

# Show graph
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
import pandas as pd

# Sample dataset
data = {
    'Hours': [1, 2, 3, 4, 5, 6, 7, 8],
    'Scores': [15, 20, 35, 40, 55, 60, 70, 85]
}

df = pd.DataFrame(data)

# Train the model
X = df[['Hours']]
y = df['Scores']

model = LinearRegression()
model.fit(X, y)

# Prediction for 3 students
for i in range(1, 4):
    hours = float(input(f"Enter study hours for Student {i}: "))
    predicted_score = model.predict([[hours]])

    print(f"Student {i} Predicted Score: {round(predicted_score[0], 2)}")
print("\nWorking prediction program.")
```
